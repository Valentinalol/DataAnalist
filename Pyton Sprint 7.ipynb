Exercici 1
Calculadora de l'índex de massa corporal
Escriu una funció que calculi l'IMC ingressat per l'usuari/ària, 
 és a dir, qui ho executi haurà d'ingressar aquestes dades. Pots obtenir més informació del seu càlcul en:
-> Índice de masa corporal IMC que es y como se calcula.
La funció ha de classificar el resultat en les seves respectives categories

```python
def get_data():
    Peso = float (input ("Cuanto pesas? Indicalo en kilos separado por un ."))
    Altura = float (input("Cuanto mides? Indicalo en metro separado por un ."))
    return Peso, Altura

def calculate_IMC (Peso, Altura):
    return Peso/(Altura**2)

def categorize_IMC (IMC):
    if IMC < 18.5:
        return "Bajo peso"
    elif IMC <= 24.9:
        return "Peso normal"
    elif IMC <= 29.9:
        return "Sobrepeso"
    else:
        return "Obesidad"
def display_result (imc_value, category):
    print (f"Tu IMC es: {imc_value:.2f} y estas en la categoria: {category}")

def main_imc():
    Peso, Altura = get_data()
    imc_value = calculate_IMC(Peso, Altura)
    category = categorize_IMC (imc_value)
    display_result (imc_value, category)

main_imc()
```

- Exercici 2
Convertidor de temperatures.

Existeixen diverses unitats de temperatura utilitzades en diferents contextos i regions. Les més comunes són Celsius (°C), Fahrenheit (°F) i Kelvin (K). També existeixen altres unitats com Rankine (°Ra) i Réaumur (°Re). Selecciona almenys 2 conversors, de tal manera que en introduir una temperatura retorni, com a mínim, dues conversions.

```python
def get_data():
    TemperaturaCelsius = float(input("Indica la Temperatura en Celsius, separada por un .: "))
    return TemperaturaCelsius

def calculoFahrenheit(TemperaturaCelsius):
    return (TemperaturaCelsius * (9/5)) + 32

def calculoKelvin(TemperaturaCelsius):
    return (TemperaturaCelsius + 273.15)

def Resultado1(Fahrenheit):
    print(f"La temperatura ingresada son {Fahrenheit:.2f} °F")

def Resultado2(Kelvin):
    print (f"La temperatura ingresada son {Kelvin:.2f} K")

def FuncionGeneral():
    TemperaturaCelsius = get_data()
    ValorFahrenheit = calculoFahrenheit(TemperaturaCelsius)
    ValorKelvin = calculoKelvin(TemperaturaCelsius)
    Resultado1(ValorFahrenheit)
    Resultado2(ValorKelvin)


FuncionGeneral()
```

- Exercici 3
Comptador de paraules d'un text.

Escriu una funció que donat un text, mostri les vegades que apareix cada paraula.

```python
from collections import Counter     #Importa Counter, cuenta elementos (como palabras) y devuelve un “diccionario de conteos”.

def get_data():
    texto = input("Escribe un texto: ")
    return texto                    #return texto devuelve esa cadena para usarla después.

def contar_palabras(texto):         #Counter(...) cuenta cuántas veces aparece cada palabra y devuelve algo parecido a {palabra: conteo}.
    return Counter(texto.split())   #texto.split() divide el texto en una lista de palabras separadas por espacios (por defecto).

def main():
    texto = get_data()
    conteo = contar_palabras(texto) #conteo = contar_palabras(texto) calcula el conteo con Counter.
    
    print("El conteo de palabras es:")
    for palabra, cantidad in conteo.items():  #recorre cada par (clave, valor) del conteo:
                                              #palabra es la palabra.
                                              #cantidad es cuántas veces aparece.
        print(f"{palabra}: {cantidad}")

main()

```

- Exercici 4
Diccionari invers.

Resulta que el client té una enquesta molt antiga que s'emmagatzema en un diccionari i els resultats els necessita al revés, és a dir, intercanviats les claus i els valors. Els valors i claus en el diccionari original són únics; si aquest no és el cas, la funció hauria d'imprimir un missatge d'advertiment.



```python
def invertir_diccionario(dic):
    dic_invertido = {}   # aquí guardaremos el nuevo diccionario
    
    # Recorremos el diccionario original
    for clave, valor in dic.items():
        # Si el valor ya existe en el diccionario invertido → es un duplicado
        if valor in dic_invertido:
            print(f"⚠️ Advertencia: el valor {valor} está duplicado, no se puede invertir.")
            return None
        
        # Si no hay problema, lo añadimos al nuevo diccionario
        dic_invertido[valor] = clave
    
    return dic_invertido


# Ejemplo sin duplicados
encuesta = {"sí": 1, "no": 2, "quizás": 3}
print(invertir_diccionario(encuesta))


# Ejemplo con duplicados
encuesta_mala = {"sí": 1, "no": 3, "quizás": 3}
print(invertir_diccionario(encuesta_mala))


```

Continuant amb l'exercici 4 del nivell 1: al client es va oblidar de comentar un detall i resulta que els valors en el diccionari original poden duplicar-se i més, per la qual cosa les claus intercanviades poden tenir duplicats. En aquest cas, en l'exercici anterior imprimies un missatge d'advertiment, ara, els valors del diccionari resultant hauran d'emmagatzemar-se com una llista. Tingues en compte que si és un valor únic no ha de ser una llista.

```python
def invertir_diccionario(dic):
    dic_invertido = {}
    
    for clave, valor in dic.items():
        if valor in dic_invertido:
            # Si no es lista todavía → convertir en lista
            if not isinstance(dic_invertido[valor], list):
                dic_invertido[valor] = [dic_invertido[valor]]
            dic_invertido[valor].append(clave)
        else:
            dic_invertido[valor] = clave
    
    return dic_invertido


# Ejemplo sin duplicados
encuesta = {"sí": 1, "no": 2, "quizás": 3}
print(invertir_diccionario(encuesta))


# Ejemplo con duplicados
encuesta_mala = {"sí": 1, "no": 3, "quizás": 3}
print(invertir_diccionario(encuesta_mala))
```

Conversió de tipus de dades

El client rep una llista de dades i necessita generar dues llistes, la primera on estaran tots els elements que es van poder convertir en flotants i l'altra on estan els elements que no es van poder convertir. Exemple de la llista que rep el client: [ '1.3', 'one' , '1e10' , 'seven', '3-1/2', ('2',1,1.4,'not-a-number'), [1,2,'3','3.4']]

```python
def listas():
    listadada = ['1.3', 'one', '1e10', 'seven', '3-1/2',
                 ('2', 1, 1.4, 'not-a-number'), [1, 2, '3', '3.4']]
    
    listafloat = []
    listaunfloat = []

    # Función interna para procesar elementos que son listas internas o tuplas
    def procesar(valor):
        if isinstance(valor, (list, tuple)):  #if lista tiene algun valor que sea lista o tupla, que reco
            for v in valor:    # recorremos sublista/tupla (Se puede obviar recorrer la tupla para evitar errores por cambios)
                procesar(v)
        else:
            try:
                f = float(valor)     # intentamos convertir a float
                listafloat.append(f)  # si funciona, lo guardo
            except (ValueError, TypeError):
                listaunfloat.append(valor) # si falla, lo guardo en otra lista

    for item in listadada: #entra en bucle de procesar
        procesar(item)

    return listafloat, listaunfloat


floats, no_floats = listas()
print("Floats:", floats)
print("No floats:", no_floats)
```
